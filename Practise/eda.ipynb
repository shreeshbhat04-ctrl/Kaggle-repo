Model

```python
from sklearn.model_selection import train_test_split

from sklearn.preprocessing import LabelEncoder, StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)
from sklearn.ensemble import RandomForestClassifier
```

```python
if "Unnamed: 13" in df.columns:
    df.drop("Unnamed: 13", axis=1, inplace=True)

print("\nAfter removing Unnamed: 13:")
print(df.shape)
```

```python
print("\n" + "=" * 80)
print("MISSING VALUES")
print("=" * 80)

print(df.isnull().sum())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print("\n" + "=" * 80)
print("TARGET VARIABLE - OUTPUT")
print("=" * 80)

print(df["Output"].value_counts())

print("\nUnique Output Values:")
print(df["Output"].unique())
```

```python
X = df.drop("Output", axis=1)
y = df["Output"]
```

```python
print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
target_encoder = LabelEncoder()
```

```python
y = target_encoder.fit_transform(y.astype(str))
```

```python
print("Target classes:")
print(target_encoder.classes_)

print("\nEncoded target:")
print(np.unique(y))
```

```python
feature_encoders = {}

for col in X.select_dtypes(include="object").columns:
    encoder = LabelEncoder()
    X[col] = encoder.fit_transform(X[col].astype(str))
    feature_encoders[col] = encoder

    print(
        f"{col} encoded -> "
        f"{len(encoder.classes_)} categories"
    )
```

```python
X = X.apply(pd.to_numeric, errors="coerce")
X = X.fillna(0)
```

```python
print("\nFeature data types:")
print(X.dtypes)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("\n" + "=" * 80)
print("TRAIN TEST SPLIT")
print("=" * 80)

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])

print(
    "Training percentage:",
    round(len(X_train) / len(X) * 100, 2),
    "%"
)

print(
    "Testing percentage:",
    round(len(X_test) / len(X) * 100, 2),
    "%"
)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model=RandomForestClassifier(n_estimators=200,random_state=42)
```

```python
results=[]
predictions={}
probabilities={}
```

```python
print("\n" + "=" * 80)
print("MODEL TRAINING")
print("=" * 80)
model.fit(X_train_scaled,y_train)
y_pred=model.predict(X_test_scaled)
y_prob=model.predict_proba(X_test_scaled)[:,1]
y_prob1=model.predict_proba(X_test_scaled)
accuracy=accuracy_score(y_test,y_pred)
precision=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred,zero_division=0)
f1=f1_score(y_test,y_pred,zero_division=0)
roc_auc=roc_auc_score(y_test,y_pred)
results.append({
        "Model": model,
        "Accuracy (%)": accuracy * 100,
        "Precision (%)": precision * 100,
        "Recall (%)": recall * 100,
        "F1 Score (%)": f1 * 100,
        "ROC-AUC (%)": roc_auc * 100
    })
predictions[model] = y_pred
probabilities[model] = y_prob
```

```python
print(y_prob)
print(y_prob1)
```

```python
results_df = pd.DataFrame(results)
```

```python
fpr, tpr, thresholds = roc_curve(
        y_test,
        y_prob
)

auc_score = roc_auc_score(
        y_test,
        y_prob
    )

plt.plot(
        fpr,
        tpr,
        linewidth=2.5,
        
)
plt.title(
    "ROC",
    fontsize=17,
    fontweight="bold"
)
show_fig()
plot_no+=1
```

```python
fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob
)

best_auc = roc_auc_score(
    y_test,
    y_prob
)

plt.figure(figsize=(10, 8))

plt.plot(
    fpr,
    tpr,
    linewidth=3,
    label=f"RandomForestClassifier (AUC = {best_auc:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=2,
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend(loc="lower right")
plt.grid(alpha=0.3)

show_fig()
plot_no += 1


```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)
plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    linewidths=1,
    cbar=False
)

plt.title(
    f"Confusion Matrix - RandomForestClassifier",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
show_fig()
plot_no+=1
```
